```python
df1.rename(columns={'sr: 0':'Innings'},inplace=True)
df1.head()
```

```python
df1.Runs=df1.Runs.apply(lambda x:x.split('*')[0])
```

```python
lst = ['DNB','TDNB']
df1[df1.Runs.isin(lst)]
```

```python
odi2=df1[~df1.Runs.isin(lst)]
odi2.Runs=odi2.Runs.astype(int)
```

```python
odi2.info()
```

```python
odi2[['4s','6s','Pos','Inns']]= odi2[['4s','6s','Pos','sr']].astype(int)
odi2['SR'] = odi2['SR'].astype(float)
odi2['Start Date'] = pd.to_datetime(odi2['Start Date'])
```

```python
odi2['Opposition'] = odi2['Opposition'].apply(lambda x:x.split('v')[-1])
odi2.info()
```
