### II) Superstore Data

In the `superstore.csv` file, you have the details of orders purchased in an American online retail store. Load the dataset, observe and analyse the different columns and answer the following questions.

```python
#Load the dataset
df2 = pd.read_csv('superstore.csv')
df2.info()
```

```python
df2.describe()
```

```python
df2["Ship Mode"].unique()
```

```python
df2["Segment"].unique()
```

```python
df2["Region"].unique()
```

```python

df2["Discount"]=df2["Discount"].apply(lambda x: float(x.strip("%")))
df2["Discount"].unique()
```

```python
df2.head()
```

```python
df3 = df2[(df2.Profit < 0) & (df2.Sales < 15000)]
sns.jointplot('Sales', 'Profit', df3)
plt.show()
```

**Q4)** Plot a pie-chart to find the Ship Mode through which most of the orders are being delivered.

     a)Standard Class
     b)First Class
     c)Second Class
     d)Same Day
     

```python
#Your code here
segment_shipmode_profit=df2.groupby(["Segment", "Ship Mode"])["Profit"].mean()
segment_shipmode_profit.plot.pie()
plt.show()
```
